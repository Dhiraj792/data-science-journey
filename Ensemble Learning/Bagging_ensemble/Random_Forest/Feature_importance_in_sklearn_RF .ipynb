```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('mnist_train.csv')
```

```python
df.head()
```

```python
X=df.iloc[:,1:]
y=df.iloc[:,0]
```

```python
sns.heatmap(X.iloc[5].values.reshape(28,28))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(X,y)
```

```python
rf.feature_importances_
```

```python
rf.feature_importances_.shape
```

```python
sns.heatmap(rf.feature_importances_.reshape(28,28))
```

# How This feature Importance Work

## To understand this we required to understand how DT will be plot

```python
from sklearn.datasets import make_classification
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
X,y=make_classification(n_samples=5,n_classes=2,n_features=2,n_informative=2,n_redundant=0,random_state=0)
```

```python
clf=DecisionTreeClassifier()
```

```python
clf.fit(X,y)
```

```python
from sklearn.tree import plot_tree
plot_tree(clf)
```

```python
clf.feature_importances_  
```
